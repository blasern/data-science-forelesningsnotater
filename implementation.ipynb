# Implementere data science løsninger

Når vi snakker om implementasjon i data science, så kan det bety flere forskjellige ting. Her er en (ikke fullstendig) liste av muligheter:

- Vi har fått noe innsikt fra data og bruker denne innsikten til å utføre en handling. Data science prosessen slutter her. Vi har for eksempel funnet ut at en ny medisin fungerer bedre enn det som ble brukt før. Vi implementerer dette ved å endre anbefalingene slik at fremtidige pasienter kan få den nye medisinen. Dette er veldig bra å få til, men det trenger vi ikke å snakke mer om.
- Vi har lagd en modell fra data og bruker den til å predikere noe. Vi lager for eksempel en nettside som bruker modellen for å forslå kundene produkter de ville likt. Det er det vi kommer til å fokusere på.
- Vi har lagd en modell, bruker den til å predikere noe og oppdaterer den når nye data blir tilgjengelig. Vi lager for eksempel en nettside som bruker modellen for å forslå kundene produkter de ville likt. Når de kjøper eller ikke kjøper genererer det nye data og vi bruker de til å oppdatere modellen vår. Dette krever metoder fra programmvareutvikling som vi ikke snakker om i dette faget.

Her gir vi bare en kort og forenklet innføring i trinnene som vanligvis inngår. I virkeligheten inngår mange flere steg som nettsteddesign, sikkerhet, oppskalering, monitoring og vedlikehold, osv.

Et minimalt eksempel som kun inneholder de mest relevante stegene er tilgjengelig i [huspris_app](https://github.com/blasern/huspris_app). 

#### Sikkerhet

Sikkerhet er en kritisk del av enhver data science-løsning som eksponeres mot brukere eller eksterne systemer. Sikkerhet er ikke noe man legger til til slutt, men noe som bør vurderes gjennom hele utviklingsprosessen. Dette inkluderer autorisering (sikre at kun autoriserte brukere dår tilgang til systemet) or databeskyttelse (ivareta sensitiv informasjon som behandles eller lagres). Det er viktig å følge gjeldende regelverk for personvern (f.eks. GDPR) særleg dersom modellen behandler personopplysninger.

#### Oppskalering

Når en data science-løsning tas i bruk med mange brukere samtidig, er det behov for oppskalering. Det vil si evnen til å handtere mange forespørsler samtidig uten at responstiden øker vesentlig eller systemet feiler.

#### Monitoring og vedlikehold

Det er avgjørende å overvåke ytelsen av en modell som er satt i produksjon over tid for å fange opp avvik eller degraderinger i prediksjonsnøyaktigheten. Et vanlig problem er såkalt konseptdrift (engl. concept drift), der forholdet med variabler og prediksjonslabel endrer seg over tid. Vedlikehold inkluderer å oppdatere modellen med ny treningsdata, validere og teste den oppdaterte modellen og håndtere programvarebiblioteker som kan endre seg. Et godt logging og varslingssystem kan være et nyttig verktøy for å oppdage problemer tidlig.
