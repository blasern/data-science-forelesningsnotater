# Data science-prosessen

Data science-prosessen består av flere steger:
- Studiedesign og datainnsamling
- Dataforberedelse og -håndtering
- Utforskende dataanalyse
- Datamodellering
- Kommunikasjon og implementasjon

Første steget er kanskje den viktigst: planlegging. Vi begynner ved å definere et presist spørsmål. Hvis vi vet ikke hva vi ønsker for å oppnå, kan vi ende opp å bruke mye tid unødvendig. Planleggingsfasen inneholder også etiske og samfunnmessige vurderinger, studiedesign, og datainnsamling.  

Neste steger er å forberede data for analysen. Dette inkluderer datahåndtering og lagring.

Selve dataanalysen består av to deler: Utforskende dataanalyse og datamodellering. I utforskende dataanalyse prøver vi å bli kjent med data. Her bruker vi statistiske metoder og visualisering. I datamodelleringsfasen lager vi prediksjonsmodeller. Her er det vanlig å bruke maskinlæring.

Dataanalysen er ikke nyttig hvis ingen vet om det eller kan ikke bruke modellene våre. Derfor må vi kommunisere funnene våre og/eller implementere løsninger. 

Vi kommer til å gå dypere inn i hver steg senere.

En ting til å legge merke til er at selv om det kan se ut at data science-prosessen er lineær er den i praksis ofte sirkulær. Det vi finner ut i ett steg fører ofte til at vi må gå tilbake og gjøre om på et tidligere steg. Kanskje viser den utforskende dataanalysen at spørsmålet vårt egentlig ikke gir mening med de dataene vi har. Kanskje viser modelleringen at vi trenger en variabel vi ikke samlet inn. Dette er ikke et tegn på at noe har gått galt, det er en naturlig og forventet del av arbeidet. Ingen klarer å planlegge et data science-prosjekt perfekt fra første forsøk, og et prosjekt der man aldri går tilbake og justerer noe, er mer sannsynlig et tegn på at man ikke har sett nøye nok etter enn et tegn på et vellykket prosjekt.

## Viktige spørsmål

Et data science prosjekt begynner ved å stille et presist spørsmål.



### Hva er egentlig spørsmålet?

Det aller første — og kanskje viktigste — steget er å finne ut nøyaktig hva vi prøver å finne ut av. Et spørsmål som ved første øyekast virker presist, viser seg ofte å skjule mange uavklarte valg. Tenk på et eksempel som å predikere eksamenskarakterene til studentene i dette kurset. Det høres konkret ut, men: Skal prediksjonen skje nå, midt i semesteret, eller rett før eksamen? Har vi tilgang til alle tidligere eksamensresultater? Kan vi be studentene ta en prøveeksam? Hvor god må prediksjonen være? Er vi fornøyde hvis vi treffer riktig karakter for halvparten av studentene? Og hvis vi bommer, er det like ille å predikere en A når fasiten er en B, som å predikere en F? Alt dette må avklares før vi i det hele tatt begynner å se på data.


### Hva slags data kan brukes for å svare på spørsmålet?

Når spørsmålet er presisert, må vi finne ut hva slags data som finnes eller kan skaffes. Det enkleste er om det allerede finnes data som er offentlig tilgjengelig, eller i det minste tilgjengelig for oss. Må vi samle inn data selv, blir prosjektet fort dyrere og tar lengre tid.

Vi må også vurdere hvor mye data vi trenger. Skal vi oppdage en sjelden bivirkning av et medikament, trenger vi svært mange pasienter. Skal vi bare finne ut om medikamentet virker i det hele tatt, holder det kanskje med noen få.

### Hva kan vi sammenligne med?

Et resultat er bare nyttig i sammenligning med et alternativ. Vi trenger derfor et utgangspunkt eller en grunnlinjemodell (*baseline*) å måle oss mot. Tenk deg at du er data science-konsulent for en nettbutikk som ønsker et anbefalingssystem. Har butikken ikke noe system fra før, er alternativet å ikke anbefale noe i det hele tatt, men en svært enkel løsning, som å anbefale de mest solgte produktene, vil sannsynligvis allerede gi en forbedring. Da bør vi sammenligne det avanserte anbefalingssystemet vårt med begge disse alternativene: ingen anbefaling, og den enkle mest-solgt-anbefalingen. Uten en slik sammenligning vet vi ikke om den kompliserte modellen faktisk er verdt innsatsen.

### Hva er de relevante etiske spørsmålene?

Til slutt må vi aldri glemme å stille de etiske spørsmålene. Hovedspørsmålet er alltid: kan dette prosjektet skade noen? Men det er nyttig å bryte dette ned i mer konkrete spørsmål.

Personvern og datainnsamling. Samler vi inn personopplysninger? Har vi lov til det, og kunne vi oppnådd det samme med anonymiserte eller aggregerte data?

Konsekvenser av feil. Hva skjer hvis modellen tar feil? I noen sammenhenger er en feilprediksjon en mindre ulempe (vi anbefaler feil bok). I andre sammenhenger, som medisinske data eller ressursfordeling i offentlig sektor, kan feil ha alvorlige konsekvenser for enkeltpersoner.

Skjevhet og diskriminering (bias og fairness). Kan modellen føre til systematisk dårligere behandling av bestemte grupper? Det klassiske eksempelet er ansettelsesmodeller: hvis et selskap historisk har ansatt flere menn enn kvinner, og vi trener en modell på disse historiske dataene, vil modellen ofte lære seg og videreføre det samme mønsteret, selv om vi aldri eksplisitt ber den ta hensyn til kjønn. Skjevhet kan komme fra dataene (datasettet representerer ikke virkeligheten godt, for eksempel en underrepresentert gruppe), eller fra modellen selv (en algoritme med antagelser som ikke passer like godt for alle grupper). Det er derfor viktig å spørre: er datasettet vårt representativt for alle gruppene modellen skal brukes på? Og gir modellen sammenlignbar ytelse på tvers av grupper, eller er den systematisk dårligere for noen?

Åpenhet og forklarbarhet. Kan vi forklare hvorfor modellen gir de prediksjonene den gir, til de som blir påvirket av beslutningen? I noen sammenhenger, spesielt der beslutninger har store konsekvenser for enkeltpersoner, som helse, forsikring eller strafferett kan det være et krav (og i EU delvis et juridisk krav under GDPR) at en person kan få en forklaring på en automatisert beslutning.

Konsekvenser av selve studiedesignet. Etiske spørsmål handler ikke bare om hva modellen gjør etterpå, men også om hvordan vi samler inn data underveis. Skal vi for eksempel undersøke om et nytt medikament virker, er det fra et rent statistisk perspektiv best å gi noen pasienter medikamentet og andre placebo. Men finnes det allerede en fungerende behandling, er det uetisk å nekte den til kontrollgruppen bare for å få et renere eksperiment.
